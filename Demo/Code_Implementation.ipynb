```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Style
sns.set_style("whitegrid")

# Data
fruits = ["Apple", "Orange", "Grapes", "Mango", "Kiwi", "Melon"]
male = np.array([40, 30, 10, 25, 35, 5])
female = np.array([30, 40, 5, 25, 10, 35])

# X positions
x = np.arange(len(fruits))
bar_width = 0.35  # width of each bar

# Create figure
plt.figure(figsize=(10, 6))

# Side-by-side bars
plt.bar(x - bar_width/2, male, width=bar_width, label="Male")
plt.bar(x + bar_width/2, female, width=bar_width, label="Female")

# Labels and title
plt.xticks(x, fruits)
plt.xlabel("Fruits")
plt.ylabel("Number of People")
plt.title("Fruit Preferences by Gender (Grouped Column Chart)")

# Legend
plt.legend()

# Optional: add value labels
for i in range(len(fruits)):
    plt.text(x[i] - bar_width/2, male[i] + 1, str(male[i]), ha='center')
    plt.text(x[i] + bar_width/2, female[i] + 1, str(female[i]), ha='center')

plt.tight_layout()
plt.show()
```

### Dual Axis Chart

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 300, 250, 400, 350]
profit = [20, 35, 30, 50, 45]

x = np.arange(len(months))

# Create figure and first axis
fig, ax1 = plt.subplots(figsize=(8, 5))

# Column chart (Sales)
ax1.bar(x, sales, label="Sales")
ax1.set_xlabel("Months")
ax1.set_ylabel("Sales")
ax1.set_xticks(x)
ax1.set_xticklabels(months)

# Second axis
ax2 = ax1.twinx()

# Line chart (Profit) → set color to orange
ax2.plot(x, profit, marker='o', color='orange', label="Profit")
ax2.set_ylabel("Profit")

# Title
plt.title("Sales (Bar) and Profit (Line) - Dual Axis Chart")

# Combine legends
lines_labels = [ax.get_legend_handles_labels() for ax in [ax1, ax2]]
lines, labels = [sum(lol, []) for lol in zip(*lines_labels)]
ax1.legend(lines, labels)

# Show plot
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

# Statistics
from scipy import stats

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid") # set style

df = pd.read_csv("titanic.csv")
# df.head()
```

### Distribution: Histogram, PMF, PDF, KDE

```python
#--------Histogram--------------
plt.hist(df['Age'].dropna(), bins=20)
plt.title("Histogram of Age")
plt.show()
```

```python
#-----------PMF (for categorical/discrete data)---------------
pmf = df['Pclass'].value_counts(normalize=True)
print(pmf)

pmf.plot(kind='bar')
plt.title("PMF of Passenger Class")
plt.ylabel("Probability")
plt.show()
```

```python
#-----------PDF (Fit data into a normal distribution)---------------
data = df['Age'].dropna()

mu, std = stats.norm.fit(data)

x = np.linspace(data.min(), data.max(), 100)
pdf = stats.norm.pdf(x, mu, std)
# print(pdf)

plt.hist(data, bins=20, density=True, alpha=0.5)
plt.plot(x, pdf)
plt.title("PDF (Normal Fit) vs Histogram")
plt.show()
```

```python
#------------KDE---------------
sns.kdeplot(data, fill=True)
plt.title("KDE of Age")
plt.show()
```

```python
sns.histplot(data, bins=20, kde=True)
plt.title("Histogram + KDE")
plt.show()
```

### Correlation Analysis

```python
#-------Scatter Plot----------
sns.scatterplot(x='Age', y='Fare', data=df)
plt.title("Age vs Fare")
plt.show()
```

```python
#----------Pearson Correlation-------
pearson_corr = df[['Age', 'Fare']].corr(method='pearson')
print("Pearson correlation:\n", pearson_corr)
```

```python
#-----------Spearman Correlation (rank-based)-------
spearman_corr = df[['Age', 'Fare']].corr(method='spearman')
print("Spearman correlation:\n", spearman_corr)
```

```python
#---------Kendall Correlation------------
kendall_corr = df[['Age', 'Fare']].corr(method='kendall')
print("Kendall correlation:\n", kendall_corr)
```

```python
#------------Heatmap Visualization-----------
corr_matrix = df[['Age', 'Fare', 'Pclass']].corr(method='spearman')

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

### Hypothesis Testing

```python
#-------------T-test (compare two groups)-----------
male_fare = df[df['Sex'] == 'male']['Fare'].dropna()
female_fare = df[df['Sex'] == 'female']['Fare'].dropna()

t_stat, p_value = stats.ttest_ind(male_fare, female_fare)

print("T-test p-value:", p_value, "T-test t-static:", t_stat)
# p < 0.05 → significant difference
```

```python
#--------------ANOVA (compare multiple groups)-----------
# fare across passenger classes
group1 = df[df['Pclass'] == 1]['Fare']
group2 = df[df['Pclass'] == 2]['Fare']
group3 = df[df['Pclass'] == 3]['Fare']

f_stat, p_value = stats.f_oneway(group1, group2, group3)

print("ANOVA p-value:", p_value)
```

```python
#-----------Chi-Square Test (categorical relationship)---------
#survival vs gender
contingency = pd.crosstab(df['Sex'], df['Survived'])

chi2, p, dof, expected = stats.chi2_contingency(contingency)

print("Chi-square p-value:", p)
```

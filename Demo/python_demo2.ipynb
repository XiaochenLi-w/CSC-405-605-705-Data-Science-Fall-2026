## Loops in Python

```python
list_ = ['hello', 'Welcome', 'Morning', 'Night']
for x in list_:
    print(x)
```

```python
for i in range(len(list_)):
    print(list_[i])
```

```python
a = 1
while a < 3:
    print('Again')
    a += 1
```

```python
array = [[1, 2, 3, 4], [5, 6, 7, 8]]
for i in range(len(array)):
    for j in range(len(array[0])):
        print(array[i][j])
```

```python
import numpy as np
a, b = np.shape(array)
for i in range(a):
    for j in range(b):
        print(array[i][j])
```

## Convert Correct Data Format

```python
import pandas as pd

data = {
  "Date": ['2020/12/01', '2020/12/02', '20201226'],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index = ['day1', 'day2', 'day3'])
print(df)
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='mixed') # avoids trying to apply a single, specific format to the entire dataset. 
# format can significantly improve performance when converting large datasets, as pandas doesn't need to infer the pattern.

print(df)
```

```python
import pandas as pd

data = {
  "Date": ['2020/12/01', '2020/12/02' , '20201226'],
  "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index = ['day1', 'day2', 'day3'])

df['Date'] = pd.to_datetime(df['Date'], format='mixed')

print(df)

df['Date'] = df['Date'].dt.strftime('%m/%d/%Y')
# dt.: operations performed on a datetime type column
# strftime: Convert the date to a string according to the specified format.

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "A": ["1", "2", "3", "bad", None]
})

print(df)

df['A'] = pd.to_numeric(df['A'], errors="coerce")
# If pandas cannot convert a value to the desired type, it will replace it with a “missing value” (NaN) instead of raising an error.

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Major": ["CS", "cs", "Computer Science", "CS", "computer science"]
})

print(df)

df["Major"] = df["Major"].str.lower()
print(df)
```

```python
major_map = {
    "cs": "Computer Science",
    "computer science": "Computer Science"
}

df["Major"] = df["Major"].replace(major_map)
print(df)
```

```python
df["Major"] = df["Major"].replace(['cs', 'computer science'], 'Computer Science')
print(df)
```

## Outliers

```python
import pandas as pd

person = {
  "name": ['Google', 'Meta', 'Apple'],
  "age": [50, 40, 12345]
}

df = pd.DataFrame(person)

df.loc[2, 'age'] = 30

print(df)
```

```python
import pandas as pd

person = {
  "name": ['Google', 'Meta', 'Apple'],
  "age": [50, 40, 12345]
}

df = pd.DataFrame(person)

for x in df.index:
  if df.loc[x, "age"] > 120:
    df.loc[x, "age"] = 120

print(df)
```

```python
import pandas as pd

person = {
  "name": ['Google', 'Meta', 'Apple'],
  "age": [50, 40, 12345]
}

df = pd.DataFrame(person)

for x in df.index:
  if df.loc[x, "age"] > 120:
      df.drop(x, inplace = True)

print(df)
```

```python
import pandas as pd

person = {
  "Name": ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
  "Age": [10, 20, 24, 26, 30, 32, 90]
}

df = pd.DataFrame(person)

Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df["Age"] < Q1 - 1.5 * IQR) |
    (df["Age"] > Q3 + 1.5 * IQR)
]
print(outliers)

df.loc[6, "Age"] = int(Q3 + 1.5 * IQR)
print(df)
```

## Remove Duplicate

```python
import pandas as pd

person = {
  "name": ['Google', 'Meta', 'Apple', 'Google'],
  "age": [50, 40, 12345, 50]
}
df = pd.DataFrame(person)

print(df.duplicated())
```

```python
import pandas as pd

person = {
  "name": ['Google', 'Meta', 'Apple', 'Google'],
  "age": [50, 40, 12345, 50]
}
df = pd.DataFrame(person)

df.drop_duplicates(inplace = True)
#df.drop_duplicates(subset=["name"], inplace = True)
print(df)
```

## In-Class Exercise

```python
import pandas as pd

df = pd.read_csv('Staff.csv')

# Handle missing values
df["Age"] = df["Age"].fillna(30)
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Remove duplicates
df = df.drop_duplicates()

# Handle outliers
median_salary = df["Salary"].median()
df.loc[df["Salary"] > 200000, "Salary"] = median_salary
df.loc[df["Age"] > 120, 'Age'] = 120

# Standardize data types
df["Age"] = df["Age"].astype("Int64")

df.to_csv('clean.csv', index = None)

print(df)
print(df.dtypes)
```

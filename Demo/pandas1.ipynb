## Arithmetic Operations with Series

```python
import pandas as pd

data = [1, 2, 3, 4, 5, 6]
index = ['a', 'b', 'c', 'd', 'e', 'f']
series = pd.Series(data, index)

print('Index:', series.index)
print('Vaues:', series.values)
print('Type:', series.dtype)
print('First two rows:', series.head(2))
```

```python
series1 = pd.Series([1, 2, 3, 4])
series2 = pd.Series([5, 6, 7, 8])

addition = series1 + series2
print(addition)
```

```python
series1 = pd.Series([1, 2, 3, 4])
series1 = series1 * 2
print(series1)
```

```python
series = pd.Series([1, 2, 3, 4])
print(series.sum(), series.mean(), series.max(), series.min(), series.std())
```

```python
filtered_series = series[series > 2]
print(filtered_series)
```

## DataFrame

```python
import pandas as pd
data = [['Google', 10], ['Meta', 20], ['Apple', 30]]

df = pd.DataFrame(data, columns = ['Company', 'Hire Number'])

print(df)
```

```python
print(df['Company'].dtype)
print(df['Hire Number'].dtype)
```

```python
df['Hire Number'] = df['Hire Number'].astype(str)
print(df['Hire Number'].dtype)
```

```python
df['Hire Number'] = df['Hire Number'].astype(float)
print(df['Hire Number'].dtype)
```

```python
import pandas as pd

data = [{'a': 1, 'b': 2}, {'a': 3, 'b': 5, 'c': 6}]
df = pd.DataFrame(data)
print(df)
```

```python
import pandas as pd

s1 = pd.Series([1, 2, 3], name="A")
s2 = pd.Series([4, 5, 6], name="B")

df = pd.concat([s1, s2], axis=1)
print(df)
```

```python
import numpy as np
import pandas as pd

ndarray_data = np.array(
    [
        ['Google', 10],
        ['Meta', 20],
        ['Apple', 30]
    ]
)

df = pd.DataFrame(ndarray_data, columns = ['Company', 'Hire Number'])
print(df)
```

```python
data = [[1, 10], [2, 20], [3, 30]]

df = pd.DataFrame(data, columns = ['Company', 'Hire Number'])

print(df.sum(), df.mean())
#print(df.shape, df.columns, df.index, df.head())
```

```python
import pandas as pd
data = [['Google', 10], ['Meta', 20], ['Apple', 30]]

df = pd.DataFrame(data, columns = ['Company', 'Hire Number'])

new_row = pd.DataFrame([['Microsoft', 40]], columns = ['Company', 'Hire Number'])
df = pd.concat([df, new_row], ignore_index = True)
print(df)
```

```python
df_dropped = df.drop(3)
print(df_dropped)
```

```python
df_dropped = df.drop(columns = ['Hire Number'])
print(df_dropped)
```

```python
df_dropped = df.drop(0, axis = 0)
print(df_dropped)

df_dropped = df.drop('Hire Number', axis = 1)
print(df_dropped)
```

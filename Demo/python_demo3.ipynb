```python
f = open('test.txt', 'r')
for line in f:
    print(line, end = '')
```

```python
f = open('test.txt', 'r')
lines = f.readlines()
print(lines)
f.close()
```

## Reading a CSV file

```python
import csv
filename = 'fast_food.csv'

csvfile = open(filename, 'r')
csvreader = csv.reader(csvfile)

fields = next(csvreader) # read header
print(fields)

rows = []
for row in csvreader:
    rows.append(row)

for i in range(5):
    print(rows[i])

print(len(rows))
```

## Reading CSV Files Into a Dictionary With csv

```python
import csv
csvfile = open(filename, 'r')
csvreader = csv.DictReader(csvfile)
rows = []
for row in csvreader:
    rows.append(row)

for i in range(5):
    print(rows[i])

print(len(rows))
```

### Writing to a csv file

```python
fields = ['Name', 'Age', 'Gender']
rows = [
    ['Bob', '20', 'Male'],
    ['Alice', '18', 'Female'],
    ['Tom', '19', 'Male']
]

filename = 'student_list.csv'
csvfile = open(filename, 'w')
csvwriter = csv.writer(csvfile)
csvwriter.writerow(fields)
csvwriter.writerows(rows)
csvfile.close()
```

### Writing a dictionary to a CSV file

```python
import csv

mydict = [{'Name': 'Bob', 'Age': '20', 'Gender': 'Male'},
         {'Name': 'Alice', 'Age': '18', 'Gender': 'Female'},
         {'Name': 'Tom', 'Age': '19', 'Gender': 'Male'}]

fields = ['Name', 'Age', 'Gender']

filename = 'student_dict_list.csv'
csvfile = open(filename, 'w')
csvwriter = csv.DictWriter(csvfile, fieldnames = fields)
csvwriter.writeheader()
csvwriter.writerows(mydict)
csvfile.close()
```

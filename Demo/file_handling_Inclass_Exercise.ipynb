```python
import csv

filename = 'fast_food.csv'
with open(filename, 'r') as csvfile:
    csvreader = csv.reader(csvfile) # step 2
    
    fields = next(csvreader)
    
    age_list = []
    for row in csvreader:
        age_list.append(int(row[0]))
    # print(age_list) # step 3
    
    max_value = max(age_list)
    min_value = min(age_list)
    avg_value = sum(age_list) / len(age_list)
    # print(max_value, min_value, avg_value) # step 4
    
    csvfile.seek(0) # very important
    
    dictreader = csv.DictReader(csvfile) # step 5
    category = {}
    for row in dictreader:
        key = row['Gender']
        if key not in category:
            category[key] = 0
        else:
            category[key] += 1
    # print(category) # step 6
    
    result_filename = 'results.txt'
    result_file = open(result_filename, 'w')
    result_file.write("Max value of Age is: " + str(max_value) + "\n")
    result_file.write("Min value of Age is: " + str(min_value) + "\n")
    result_file.write("Average value of Age is: " + str(avg_value) + "\n")
    result_file.write(str(category) + "\n")
```

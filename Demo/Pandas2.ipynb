```python
import pandas as pd

df = pd.read_excel('runoob_pandas_data.xlsx', sheet_name = 'Sheet2')
print(df)
```

```python
dfs = pd.read_excel('runoob_pandas_data.xlsx', sheet_name = ['Sheet1', 'Sheet2'])
print(dfs)
print(dfs['Sheet1']['City'])
```

## CSV Write

```python
import pandas as pd

nme = ['A', 'B', 'C', 'D']
st = ['Apple', 'Watermellon', 'Peach', 'Pear']
ag = [90, 80, 100, 120]

dict = {'nme': nme, 'site': st, 'age': ag}

df = pd.DataFrame(dict)

df.to_csv('output.csv', index = False)
```

```python
import pandas as pd

df = pd.DataFrame({
'Name': ['Alice', 'Bob', 'Charlie'],
'Age': [25, 30, 35],
'City': ['New York', 'Los Angeles', 'Chicago']
})

df.to_excel('output.xlsx', sheet_name='Sheet1', index=False)
```

```python
with pd.ExcelWriter('output.xlsx') as writer:
    df.to_excel(writer, sheet_name='Sheet1', index=False)
    df.to_excel(writer, sheet_name='Sheet2', index=False)
```

## In-class Exercise

```python
import pandas as pd

# -------------------------
# Part A: Read the CSV file
# -------------------------

# 1. Read the CSV file
df = pd.read_csv("titanic.csv")

# 2. Display the first 6 rows
print("First 6 rows:")
print(df.head(6))

# 3. Show shape, columns, and data types
print("\nShape of DataFrame:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)


# -----------------------------------
# Part B: DataFrame and Series operations
# -----------------------------------

# 1. Age column statistics
age_series = df["Age"]

print("\nMean age:")
print(age_series.mean())

print("\nMaximum age:")
print(age_series.max())

# 2. Count passengers in each class
print("\nPassenger count by class:")
print(df["Pclass"].value_counts())

# 3. Filter passengers with age > 30
age_over_30 = df[df["Age"] > 30]
print("\nPassengers with age > 30:")
print(age_over_30.head())

# 4. Average age of survivors
survivors = df[df["Survived"] == 1]
print("\nAverage age of survivors:")
print(survivors["Age"].mean())


# -------------------------
# Part C: Excel operations
# -------------------------

# 1. Save to Excel
df.to_excel("titanic.xlsx", index=False)
print("\nData saved to titanic.xlsx")

# 2. Read from Excel
df_excel = pd.read_excel("titanic.xlsx")

# 3. Display first 3 rows
print("\nFirst 3 rows from Excel file:")
print(df_excel.head(3))
```
